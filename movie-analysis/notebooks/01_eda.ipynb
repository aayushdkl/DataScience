```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Load the Dataset
Load the raw movie ratings dataset from the `data/raw/` folder.

```python
# Load the dataset
df = pd.read_csv('../data/raw/movies.csv')
df.head()  # Display first 5 rows
```

## 2. Basic Exploration
Check the dataset's structure and summary statistics.

```python
# Check data types and missing values
print(df.info())

# Summary statistics
print(df.describe())

# Unique genres
print(df['genre'].unique())
```

## 3. Visualizations
Visualize the distribution of ratings and popular genres.

```python
# Plot average rating by genre
plt.figure(figsize=(10, 6))
sns.barplot(x='genre', y='rating', data=df)
plt.title('Average Rating by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Rating')
plt.show()

# Plot rating distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['rating'], bins=10, kde=True)
plt.title('Distribution of Movie Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```
